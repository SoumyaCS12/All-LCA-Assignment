```python
side1 = float(input("Enter the length of the first side of the triangle:"))
side2 = float(input("Enter the length of the second side of the triangle:"))
side3 = float(input("Enter the length of the third side of the triangle:"))

if side1**2 + side2**2 == side3**2:
    print("Right Triangle!")
elif side1**2 + side3**2 == side2**2:
    print("Right Triangle!")
elif side2**2 + side3**2 == side1**2:
    print("Right Triangle!")
else:
    print("Not a right Triangle")
```

```python
#1
def prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return factors

num = int(input("Input number = "))
print("prime factors =", ",".join(map(str, prime_factors(num))))
```

```python
#2
def even (start, end):
    for i in range(start, end + 1):
        if i % 2 == 0:
            print(i, end=" ")

even(1,100)
```

```python
#3
import math

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

discriminant = b**2 - 4*a*c

if discriminant < 0:
    print("The equation has no real roots (discriminant is negative).")

elif discriminant == 0:
    root1 = -b / (2*a)
    print("The equation has one real root (both roots are equal).")
    print("Root =", root1)

else:
    sqrt_val = math.sqrt(discriminant)
    root1 = (-b + sqrt_val) / (2*a)
    root2 = (-b - sqrt_val) / (2*a)
    print("The equation has two different real roots.")
    print("Root 1 =", root1)
    print("Root 2 =", root2)
```
