```python
side1 = float(input("Enter the length of the first side of the triangle:"))
side2 = float(input("Enter the length of the second side of the triangle:"))
side3 = float(input("Enter the length of the third side of the triangle:"))

if side1**2 + side2**2 == side3**2:
    print("Right Triangle!")
elif side1**2 + side3**2 == side2**2:
    print("Right Triangle!")
elif side2**2 + side3**2 == side1**2:
    print("Right Triangle!")
else:
    print("Not a right Triangle")
```

```python
#1
def prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return factors

num = int(input("Input number = "))
print("prime factors =", ",".join(map(str, prime_factors(num))))
```

```python
#2
def even (start, end):
    for i in range(start, end + 1):
        if i % 2 == 0:
            print(i, end=" ")

even(1,100)
```

```python
#3
import math

def quadratic_roots(a, b, c):
    try:
        discriminant = b**2 - 4*a*c
        if discriminant < 0:
            raise ValueError("math domain error")
        root1 = (-b + math.sqrt(discriminant)) / (2*a)
        root2 = (-b - math.sqrt(discriminant)) / (2*a)
        return root1, root2
    except ValueError as ve:
        print("ValueError:", ve)
    except NameError as ne:
        print("NameError:", ne)
    except TypeError as te:
        print("TypeError:", te)

try:
    a = float(input("Enter coefficient a: "))
    b = float(input("Enter coefficient b: "))
    c = float(input("Enter coefficient c: "))
    result = quadratic_roots(a, b, c)
    if result:
        print("Roots are:", result)
except ValueError:
    print("ValueError: Please enter valid numbers")
```
