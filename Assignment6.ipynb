```python
#Remove Rows

import pandas as pd
import numpy as np
file = pd.read_excel(r"C:\Users\Admin\Downloads\PNRao_Universal_Basic_Employee_Data.xlsx")
#Drop NA files - take out the row completely.
# By default, the dropna() method returns a new DataFrame, and will not change the original.
new_file = file.dropna()
print(new_file.to_string())

#to change the original DataFrame, use the (inplace = True)
file.dropna(inplace = True)
print(file.to_string())
```

```python
#Replace Empty Values

#Replace NULL values with the number 130
file.fillna(130, inplace = True)

#Replace NULL values in the "LastName" columns with the number 130:
file.fillna({"Salary": 130}, inplace=True)
```

```python
#Convert Data of Wrong Format Into a Correct Format

#to_datetime()
file['HireDate'] = pd.to_datetime(file['HireDate'], format='mixed')
print(file.to_string())
```

```python
#Removing Rows - Remove rows with a NULL value in the "Date" column:
file.dropna(subset=['HireDate'], inplace = True)
```

```python
#Replacing Values

#If the value is higher than 100000, set it to 100000:
for x in file.index:
  if file.loc[x, "Salary"] > 100000:
    file.loc[x, "Salary"] = 100000
      
print(file.to_string())


#Delete rows where "Salary" is higher than 100000:
for x in file.index:
  if file.loc[x, "Salary"] > 100000:
    file.drop(x, inplace = True)

print(file.to_string())
```

```python
#Discovering Duplicates

#Returns True for every row that is a duplicate, otherwise False:
print(file.duplicated())

#Removing Duplicates
file.drop_duplicates(inplace = True)
```
