```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np

df = pd.read_csv("Assignment7.csv")
print("Dataset: \n", df)
```

```python
#line graph
plt.figure(figsize=(7,4.5))
plt.plot(df["firstname"], df["python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
#bar plot
plt.figure(figsize=(7,4.5))
plt.bar(df["firstname"], df["Math"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.grid(False) 
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(7,4.5))
plt.scatter(df["firstname"], df["python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
#Correlation matrix  
correlation = df[["Math", "python", "Attendance"]].corr()
print("Correlation Matrix")
print(correlation)
```

```python
#Heatmap
plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap='Blues', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

```python
#Pie Chart
y = df["python"]

plt.pie(y, labels = df["firstname"])
plt.title("Python Marks of Students")
plt.show() 
```
