```python
#7
String = "python batch one"
freq = {}

for char in String:
    if char.isalpha():                 
        char = char.lower()            
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

print("original string:", String)
print("frequency:")
print(freq)
```

```python
#9
from datetime import date 
d1 = int(input("enter the day for date one:"))
m1 = int(input("enter the month for date one:"))
y1 = int(input("enter the year for date one:"))
date1 = (d1, m1, y1)

d2 = int(input("enter the day for date two:"))
m2 = int(input("enter the month for date two:"))
y2 = int(input("enter the year for date two:"))
date2 = (d2, m2, y2)

date11 = date(date1[2], date1[1], date1[0])   
date22 = date(date2[2], date2[1], date2[0])
difference = date22 - date11

print("Date 1:", date1)
print("Date 2:", date2)
print("Number of days between the two dates:", abs(difference.days))
```
