```python
import re
txt = "The rain in Spain"
x = re.search("^The.*Spain$", txt)
if x:
    print("Yes! We have a match!")
else:
    print("No match")
```

```python
import re 
def check_string(s):
    pattern = "^[A-Za-z0-9]+$"
    return bool(re.match(pattern,s))

s1 = "Hello123"
s2 = "Hello_123"

print(check_string(s1))
print(check_string(s2))

#----------------------------------------------

def check_string(s):
    return s.isalnum()

s1 = "Hello123"
s2 = "Hello123!"

print(check_string(s1))
print(check_string(s2))
```
