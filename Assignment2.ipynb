```python
#largest number 
num1 = float(input("Input your first number: "))
num2 = float(input("Input your second number: "))
num3 = float(input("Input your third number: "))

ListMarks = [num1, num2, num3]
print(ListMarks)

if ListMarks[0] < ListMarks[1]:
    if ListMarks[1] < ListMarks[2]:
            print(ListMarks[2])
    else:
            print(ListMarks[1])
else: 
        if ListMarks[0] < ListMarks[2]:
                print(ListMarks[2])
        else:
                print(ListMarks[0])
```

```python
#FAQ Marks in Math
marks = float(input("Enter your grade from Math: "))

if marks >= 90 and marks <= 100:
    print("Grade: O")
elif marks >= 80 and marks <= 89:
    print("Grade: A+")
elif marks >= 70 and marks <= 79:
    print("Grade: A")
elif marks >= 60 and marks <= 69:
    print("Grade: B")
elif marks >= 50 and marks <= 59:
    print("Grade: C")
elif marks >= 40 and marks <= 49:
    print("Grade: P")
elif marks >= 0 and marks <= 39:
    print("Grade: F")
else:
    print("Invalid number of marks have been entered")
```
